# Natural Language Processing (NLP) Fundamentals

* **Natural Language Processing (NLP):** Methods for turning language into representations a computer can work with, then using those representations to perform a task.
* **Token:** One occurrence produced by a tokenizer. The same type can appear as many tokens.
* **Type · vocabulary:** A type is one distinct token form; the vocabulary is the set of types kept by the representation.
* **Tokenization:** The decision about where one text unit ends and the next begins. Different tokenizers can reasonably split the same string differently.
* **Subword tokenization:** Representing text with reusable pieces smaller than whole words. Modern language models use fixed subword vocabularies so unfamiliar words can still be represented.
* **Token ID:** The integer index used to look up a token or subword piece in a model's vocabulary.
* **Preprocessing:** Choices made before measurement or modeling, such as lowercasing, lemmatization, punctuation filtering, and stopword removal. Every one can remove useful information.
* **Stopword:** A very frequent function word, such as *the*, *of*, or *in*, that many analyses drop to leave the content words.
* **Lemmatization:** Mapping each word form to its dictionary form, its lemma (*heroes* → *hero*, *were* → *be*).
* **Frequency · relative frequency:** A raw count of how often a type occurs; relative frequency divides by the total number of tokens so differently sized corpora can be compared more sensibly.
* **Hapax:** A type observed exactly once in the corpus.
* **Zipf's law:** The recurring empirical pattern that word frequency falls roughly as an inverse power of frequency rank. A straight-ish log-log plot is suggestive, not a proof that a corpus follows an exact Zipf model.
* **Concordance / KWIC:** The occurrences of a target word shown with a small window of surrounding text. It answers: *how is this word used here?*
* **similar():** An NLTK operation that looks for words occupying comparable immediate contexts. Similar context does not guarantee synonymy.
* **N-gram:** A sequence of neighboring tokens. Unigrams keep one token, bigrams two, trigrams three; longer n-grams keep more local order and become sparser.
* **Collocation:** A pair or sequence that occurs together more strongly than its individual word frequencies alone would suggest.
* **Document:** The unit of text you choose to represent, for example one Wikipedia page, article, speech, review, or user's posts.
* **Bag of Words:** A document representation that counts vocabulary terms while discarding word order.
* **Document-term matrix:** Rows are documents, columns are vocabulary terms, and each cell contains that term's value for that document (here, a count).
* **Sparse matrix:** A matrix in which most entries are zero, as happens when a large vocabulary is shared by documents that each use only a small part of it.
