# Essentials

#### Walk, path, shortest path
A walk is any sequence of linked nodes; a path never repeats a node; the shortest path between two nodes has the fewest links, and its length is their __distance__ $d_{ij}$ (infinite if no path exists).

#### Breadth-first search (BFS)
From a start node, visit all nodes at distance 1, then all unvisited nodes at distance 2, and so on. The ring a node lands in is its distance. One run gives the distances from the start to every node.

#### Eccentricity, diameter, radius, and center
A node's eccentricity is its distance to the farthest node; the diameter is the largest eccentricity, the radius the smallest, and the center the node(s) attaining it. Marvel: diameter 6, radius 3, center Spider-Man.

#### Reachability in directed networks
Distances follow the arrows, so $d_{AB}$ and $d_{BA}$ can differ, and one can be infinite. __Weakly connected__: connected if you ignore direction. __Strongly connected__: every node reaches every other following the arrows.

#### Degree centrality
$k_i$; in directed networks $k_i^{in}$ (who is mentioned) and $k_i^{out}$ (who mentions) separately. Local and cheap. The question: _how many connections?_

#### Closeness centrality
$C_i = (n - 1) / \sum_{j \neq i} d_{ij}$, the inverse mean distance to everyone. High = everything is near. Breaks (goes to 0) if any node is unreachable. The question: _how fast can you reach everyone?_

#### Harmonic centrality
$\frac{1}{n-1} \sum_{j \neq i} 1/d_{ij}$ with $1/\infty = 0$. Closeness that survives disconnected and directed networks.

#### Betweenness centrality
$b_i = \sum_{s \neq i \neq t} \sigma_{st}(i)/\sigma_{st}$, the summed fraction of shortest paths between other pairs that pass through $i$; normalized by the number of pairs. High = a __broker__ or bridge; can be high at low degree (the only door to a region) and near zero at high degree (neighbors all linked to each other). The question: _who controls the flow?_

#### Eigenvector centrality
$x_i = \frac{1}{\lambda} \sum_j A_{ij} x_j$: a node is important if its neighbors are important. The leading eigenvector of the adjacency matrix; computed by iterating "replace each score by the sum of the neighbors' scores, rescale."

#### PageRank
The directed version with a random walker: $PR_i = \frac{1-\alpha}{n} + \alpha \sum_{j \to i} PR_j / k_j^{out}$. Each page splits its vote among its out-links; with probability $1-\alpha$ the walker __teleports__ to a random page, which fixes dangling pages and traps. $\alpha = 0.85$ by convention. The question: _who is linked to by the important?_

#### Random walk and stationary distribution
On an undirected network a random walker spends time in proportion to degree, $\pi_i = k_i / 2m$. Random-walk importance is degree unless direction is involved.

#### Centrality, compared to what
A centrality is only meaningful relative to a null. Against the degree-preserving shuffle, a node whose betweenness matches its shuffled value is "just its degree"; the surprises ($|z|$ large) are the brokers and the sheltered.

#### Degree assortativity $r$
The Pearson correlation between the degrees at the two ends of a link, over all links. $r > 0$ __assortative__ (hubs link to hubs, typical of social networks), $r < 0$ __disassortative__ (hubs link to leaves, typical of technological and biological ones). A star has $r = -1$.

#### Average neighbor degree $k_{nn}(k)$
The mean degree of the neighbors of nodes with degree $k$. Rising = __assortative__, falling = __disassortative__, flat at $\langle k^2 \rangle / \langle k \rangle$ = no mixing.

#### Structural disassortativity
In a heavy-tailed network the hubs must link mostly to small nodes (there are not enough hubs), so $r < 0$ appears even under the degree-preserving null. Marvel: $-0.105$ real vs $-0.118 \pm 0.009$ shuffled. Compare to the null before calling a network disassortative.

#### Homophily
Assortative mixing by an attribute (team, age, opinion) rather than degree: like links to like. Measured by the attribute assortativity; tested by shuffling the labels across nodes.

#### Clique, maximal clique, and clique number
A set of nodes in which every pair is linked; a $k$-clique has $k(k-1)/2$ links. Maximal = no node can be added. The clique number is the size of the largest clique. A 3-clique is a triangle. Marvel's clique number is 8.
